```python
# Python version (index starts at 0)
x = [1, 2, 3]    # Input signal
h = [4, 5]       # Filter

N = len(x)
M = len(h)
L = N + M - 1    # Output length

y = [0] * L      # Preallocate output

for n in range(L):
    for k in range(M):
        if 0 <= n - k < N:
            y[n] += x[n - k] * h[k]

print("Python output:")
print(y)
```

```python
# Python convolution with 1-based indexing (MATLAB-style)

# Original input signal and filter
x = [1, 2, 3]      # length N = 3
h = [4, 5]         # length M = 2

# Simulate 1-based indexing by padding
x = [0] + x        # x[1] = 1, x[2] = 2, x[3] = 3
h = [0] + h        # h[1] = 4, h[2] = 5

N = len(x) - 1     # actual signal length
M = len(h) - 1     # actual filter length
L = N + M          # full convolution length

# Initialize output with 1-based indexing
y = [0] * (L + 1)  # y[1] to y[L]

# Convolution loop (mimicking MATLAB)
for n in range(1, L + 1):
    for k in range(1, M + 1):
        if 1 <= n - k + 1 <= N:
            y[n] += x[n - k + 1] * h[k]

# Remove dummy 0-th entry
y = y[1:]

print("Convolution result with 1-based indexing:")
print(y)
```

```python
# Discrete convolution with 0-based indexing
# No 'if' inside the loop — loop bounds are explicitly managed

x = [1, 2, 3]    # Input signal (length N = 3)
h = [4, 5]       # Filter (length M = 2)

N = len(x)
M = len(h)
L = N + M - 1    # Length of the full convolution output

y = [0] * L      # Output signal

# Outer loop over output samples
for n in range(L):
    # Calculate valid k range to avoid out-of-bounds
    k_min = max(0, n - (N - 1))
    k_max = min(M - 1, n)

    for k in range(k_min, k_max + 1):
        y[n] += x[n - k] * h[k]

print("Convolution result (0-based, no if):")
print(y)
```
